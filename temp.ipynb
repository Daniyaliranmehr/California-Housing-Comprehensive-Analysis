```python
import numpy as np
import matplotlib.pyplot as plt

delta = 1.0

error = np.linspace(-4, 4, 1000)

c = 1.0  # scale parameter

# Log-Cosh Loss
f = np.log(np.cosh(error / c))

plt.figure(figsize=(8, 5), facecolor="#1e1f22")

ax = plt.gca()
ax.set_facecolor("#1e1f22")

plt.plot(error, f, linewidth=2)

plt.title("Log-Cosh Loss", color="white")
plt.xlabel("Prediction Error", color="white")
plt.ylabel("Loss", color="white")

plt.tick_params(colors="white")
plt.grid(True, alpha=0.3)

plt.savefig("E:/Deep_learning/Powerpoints/", dpi=300, bbox_inches='tight')
plt.show()
```
