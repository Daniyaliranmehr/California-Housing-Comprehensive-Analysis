```python
import numpy as np
import matplotlib.pyplot as plt

error = np.linspace(-4, 4, 1000)
c = 1.0

quadratic = 0.5 * (error / c) ** 2
charbonnier = np.sqrt((error / c) ** 2 + 1) - 1
cauchy = np.log(0.5 * (error / c) ** 2 + 1)
geman_mcclure = 2 * (((error / c) ** 2 / 4) / (1 + (error / c) ** 2 / 4))
welsch = 1 - np.exp(-0.5 * (error / c) ** 2)

plt.figure(figsize=(10, 6), facecolor="#1e1f22")
ax = plt.gca()
ax.set_facecolor("#1e1f22")

plt.plot(error, quadratic, linewidth=2, label="Quadratic (α = 2)")
plt.plot(error, charbonnier, linewidth=2, label="Charbonnier (α = 1)")
plt.plot(error, cauchy, linewidth=2, label="Cauchy (α = 0)")
plt.plot(error, geman_mcclure, linewidth=2, label="Geman-McClure (α = -2)")
plt.plot(error, welsch, linewidth=2, label="Welsch (α → -∞)")

plt.title("Adaptive Loss Functions", color="white")
plt.xlabel("Prediction Error", color="white")
plt.ylabel("Loss", color="white")

plt.tick_params(colors="white")
plt.grid(True, alpha=0.3)

legend = plt.legend()
for text in legend.get_texts():
    text.set_color("#1e1f22")

plt.savefig(
    "E:/Deep_learning/Powerpoints/Adaptive_Loss_Comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
