# Test the Best Model on the Test Set

In this section, the best model (`AdamW_best_model.pth`), trained with the MAE loss function and the AdamW optimizer, is used to make predictions on the test set and evaluate its final performance.

## Libraries

```python
# Prevent OpenMP runtime conflict between libraries like PyTorch, TensorFlow, and NumPy on Windows
import os
os.environ["KMP_DUPLICATE_LIB_OK"] = "TRUE"

import sys
import os
sys.path.append(os.path.abspath(".."))

import importlib
import src.training_utils as tu
importlib.reload(tu)

import torch
```

## Load Data

Load the test set prepared in `01_data_preparation.ipynb`.

```python
data = torch.load('../saved_values/data.pt', weights_only=False)

x_test = torch.FloatTensor(data["x_test"])
y_test = torch.FloatTensor(data["y_test"])

print(f"x_test shape: {x_test.shape}\ny_test shape: {y_test.shape}")
```

## Load Model

Load the best-performing model (`AdamW_best_model.pth`), trained with MAE and AdamW

```python
model = tu.Model(in_features=20, h1=64, h2=32, out_features=1)

model.load_state_dict(torch.load("../saved_values/AdamW_best_model.pth"))
```

## Prediction

```python
loss_fn = torch.nn.L1Loss()
```

```python
evaluator = tu.Evaluator(model, loss_fn)

results = evaluator.test(x_test, y_test)
```

```python
print(f"R2 Score: {results["R2"]:.4f}\nLoss: {results["Loss"]:.4f}")
```
